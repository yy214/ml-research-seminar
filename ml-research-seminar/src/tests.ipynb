```python
from ollama import chat
```

```python
class LlamaChat:
  def get_response(self, messages):
    response = chat(
        model="qwen2-math:7b",
        messages=messages,
    )
    return response.message.content
```

```python
lc = LlamaChat()

user_input = "I have 2 apples in hand, Jimmy gives me 4 more, and I give 3 to Anna. How much apples do I have left?"
lc.get_response(messages=[{"role": "user", "content": user_input}])
```

```python
# SIMPLE TESTING

print("Llama LLM Chat - Type 'exit' to quit.")

lc = LlamaChat()
messages = []

while True:
    user_input = input("\nYou: ")

    if user_input.lower() in ["exit", "quit"]:
        print("\nExiting. Goodbye!")
        break

    messages.append({"role": "user", "content": user_input})
    response = lc.get_response(messages)
    messages.append({"role": "assistant", "content": response})
    print(f"Llama: {response}")
```
